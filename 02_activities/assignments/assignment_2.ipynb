```python
# Your answer here
# Manual example: 
# Input: [3, 4, 2, 5, 6, 2, 3, 3, 1]
# Output: [[3, 4, 5, 3], [3, 4, 5, 1], [3, 4, 6], [3, 2, 2], [3, 2, 3]]

#Example1 - From Parisa's assignment
root = TreeNode(5)
root.left = TreeNode(3)
root.right = TreeNode(8)
root.left.left = TreeNode(1)
root.left.right = TreeNode(4)

print(bt_path(root))

# In this approach, each value of the tree is defined through the specific node they they belong to, if using the same convention as the examples in the assignment, the input would be equivalent to: [5, 3, 8, 1, 4]

# The output was achieved by running the input through the program, and printing the result, which is: [[5, 3, 1], [5, 3, 4], [5, 8]].
```


-   Copy the solution your partner wrote. 


```python
# Your answer here

# Definition for a binary tree node.
class TreeNode(object):
     def __init__(self, val = 0, left = None, right = None):
         self.val = val
         self.left = left
         self.right = right
def bt_path(root: TreeNode) -> list[list[int]]:
  def dfs(node, path, paths):
        if not node:
            return
        path.append(node.val)
        
        if not node.left and not node.right:
            paths.append(list(path))
        else:
            dfs(node.left, path, paths)
            dfs(node.right, path, paths)
        
        path.pop()
    
  paths = []
  dfs(root, [], paths)
  return paths
```


-   Explain why their solution works in your own words.


```python
# Your answer here
# The solution works through recursion.
# The function is given a TreeNode as input, and the dfs function would recursively
# access each element of the node going deeper and deeper with each pass,
# the dfs function also takes in two additional parameters, path and paths.
# path is initialized with an empty list, and paths is just defined as a variable.
# As the program goes down through each path, the value of the node is stored in 
# path, and once it has gone through the last element, path is appended to paths.
# After going through all possible paths, the function would then return paths
# which contain the final output.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Your answer here
# Time complexity:
# Worst case: O(n log n)
# The worst case would be by having a balanced binary tree as the input,
# where each node would have a value, each pass, halving the number of elements left.
# Best case: O(n)
# The best case is where we have a skewed binary tree, where in only one side is
# filled, this would be essentially traversing through n elements O(n)
# 
# Space complexity:
# Best case: O(n)
# In a skewed tree, there would be O(n) space complexity, to traverse through one path
# Worst case: O(n log n)
# In a balanced tree: 
# There would be O(n/2) paths, which could be simplified to O(n)
# Plus the space of the call stack which would be the height O(h) where h is log n
```
